```python
# نگاشت برای بی‌خوابی (۵ سطح)
insomnia_map = {
    'هرگز': 0,
    'کمی': 1,
    'متوسط': 2,
    'شدید': 3,
    'بسیار شدید': 4
}

# نگاشت برای اضطراب (۴ سطح)
anxiety_map = {
    'اصلا': 0,
    'چندین روز': 1,
    'بیش از نیمی از روز': 2,
    'تقریبا هر روز': 3
}

# نگاشت برای نشخوار فکری (۵ سطح)
rumination_map = {
    'خیلی مخالفم': 0,
    'مخالفم': 1,
    'نه موافقم و نه مخالفم': 2,
    'موافقم': 3,
    'خیلی موافقم': 4
}

# اعمال نگاشت روی هر ستون
for col in insomnia_cols:
    df[col] = df[col].map(insomnia_map)

for col in anxiety_cols:
    df[col] = df[col].map(anxiety_map)

for col in rumination_cols:
    df[col] = df[col].map(rumination_map)

print("✅ کدگذاری انجام شد.")
```

```python
# تحلیل میانجی با استفاده از رگرسیون (روش بارون و کنی)
import statsmodels.api as sm

# مسیر a: اضطراب → نشخوار
X = df['نمره_اضطراب']
M = df['نمره_نشخوار']
X = sm.add_constant(X)
model_a = sm.OLS(M, X).fit()
print("مسیر a (اضطراب → نشخوار):", model_a.params[1], "p=", model_a.pvalues[1])

# مسیر b: نشخوار → بی‌خوابی (با کنترل اضطراب)
X = df[['نمره_اضطراب', 'نمره_نشخوار']]
X = sm.add_constant(X)
Y = df['نمره_بیخوابی']
model_b = sm.OLS(Y, X).fit()
print("مسیر b (نشخوار → بی‌خوابی با کنترل اضطراب):", model_b.params[2], "p=", model_b.pvalues[2])

# مسیر c (کل): اضطراب → بی‌خوابی
X = df['نمره_اضطراب']
X = sm.add_constant(X)
model_c = sm.OLS(Y, X).fit()
print("مسیر کل c (اضطراب → بی‌خوابی):", model_c.params[1], "p=", model_c.pvalues[1])

# مسیر c' (مستقیم): اضطراب → بی‌خوابی با کنترل نشخوار
print("مسیر مستقیم c' (اضطراب → بی‌خوابی با کنترل نشخوار):", model_b.params[1], "p=", model_b.pvalues[1])
```

```python
import statsmodels.api as sm
import pandas as pd

# 1️⃣ حذف ردیف‌های دارای مقدار گمشده در متغیرهای اصلی
df_clean = df[['نمره_بیخوابی', 'نمره_اضطراب', 'نمره_نشخوار']].dropna()

# 2️⃣ مسیر a: اضطراب → نشخوار
X = df_clean['نمره_اضطراب']
M = df_clean['نمره_نشخوار']
X = sm.add_constant(X)
model_a = sm.OLS(M, X).fit()
print("مسیر a (اضطراب → نشخوار):")
print(f"ضریب = {model_a.params['نمره_اضطراب']:.3f}, p-value = {model_a.pvalues['نمره_اضطراب']:.3f}")

# 3️⃣ مسیر b: نشخوار → بی‌خوابی (با کنترل اضطراب)
X = df_clean[['نمره_اضطراب', 'نمره_نشخوار']]
X = sm.add_constant(X)
Y = df_clean['نمره_بیخوابی']
model_b = sm.OLS(Y, X).fit()
print("\nمسیر b (نشخوار → بی‌خوابی با کنترل اضطراب):")
print(f"ضریب = {model_b.params['نمره_نشخوار']:.3f}, p-value = {model_b.pvalues['نمره_نشخوار']:.3f}")

# 4️⃣ مسیر کل c: اضطراب → بی‌خوابی (بدون میانجی)
X = df_clean['نمره_اضطراب']
X = sm.add_constant(X)
model_c = sm.OLS(Y, X).fit()
print("\nمسیر کل c (اضطراب → بی‌خوابی):")
print(f"ضریب = {model_c.params['نمره_اضطراب']:.3f}, p-value = {model_c.pvalues['نمره_اضطراب']:.3f}")

# 5️⃣ مسیر مستقیم c': اضطراب → بی‌خوابی با کنترل نشخوار
print("\nمسیر مستقیم c' (اضطراب → بی‌خوابی با کنترل نشخوار):")
print(f"ضریب = {model_b.params['نمره_اضطراب']:.3f}, p-value = {model_b.pvalues['نمره_اضطراب']:.3f}")

# 6️⃣ (اختیاری) آزمون سوبل – با کتابخانه pingouin
# !pip install pingouin  # اگر نصب نیست
# import pingouin as pg
# sobel = pg.sobel(df_clean['نمره_اضطراب'], df_clean['نمره_نشخوار'], df_clean['نمره_بیخوابی'])
# print("\nآزمون سوبل:", sobel)
```

```python
from scipy.stats import norm
import numpy as np

# استخراج ضرایب و خطای استاندارد از مدل‌های قبلی
# مسیر a (اضطراب -> نشخوار)
coef_a = model_a.params['نمره_اضطراب']
se_a = model_a.bse['نمره_اضطراب']

# مسیر b (نشخوار -> بی‌خوابی با کنترل اضطراب)
coef_b = model_b.params['نمره_نشخوار']
se_b = model_b.bse['نمره_نشخوار']

# محاسبه اثر غیرمستقیم (Indirect effect)
indirect_effect = coef_a * coef_b

# فرمول سوبل برای محاسبه خطای استاندارد اثر غیرمستقیم
sobel_se = np.sqrt((coef_b**2 * se_a**2) + (coef_a**2 * se_b**2))

# محاسبه آماره Z
sobel_z = indirect_effect / sobel_se

# محاسبه p-value (دو دنباله)
p_value = 2 * (1 - norm.cdf(abs(sobel_z)))

# چاپ نتایج
print("📊 نتایج آزمون سوبل (Sobel Test):")
print(f"اثر غیرمستقیم (a × b) = {indirect_effect:.4f}")
print(f"خطای استاندارد اثر غیرمستقیم = {sobel_se:.4f}")
print(f"آماره Z = {sobel_z:.3f}")
print(f"p-value = {p_value:.4f}")

if p_value < 0.05:
    print("✅ نتیجه: اثر غیرمستقیم در سطح 0.05 معنادار است (میانجی‌گری تأیید می‌شود).")
else:
    print("❌ نتیجه: اثر غیرمستقیم در سطح 0.05 معنادار نیست (نقش میانجی رد می‌شود).")
```

```python
import pandas as pd
import numpy as np
from scipy.stats import pearsonr

# ===================================================
# 1️⃣ جدول فراوانی جمعیت‌شناختی (ترکیبی)
# ===================================================
gender_freq = df['جنسیت شما کدام است؟'].value_counts().reset_index()
gender_freq.columns = ['جنسیت', 'تعداد']

edu_freq = df['تحصیلات شما کدام است؟'].value_counts().reset_index()
edu_freq.columns = ['تحصیلات', 'تعداد']

age_freq = df['گروه سنی شما کدام است؟'].value_counts().reset_index()
age_freq.columns = ['گروه سنی', 'تعداد']

# ترکیب سه جدول در یک دیتافریم (برای نمایش بهتر)
demographic_table = pd.concat([
    gender_freq.rename(columns={'تعداد': 'جنسیت'}),
    edu_freq.rename(columns={'تعداد': 'تحصیلات'}),
    age_freq.rename(columns={'تعداد': 'گروه سنی'})
], axis=1)

print("📊 جدول ۱: توزیع فراوانی جمعیت‌شناختی")
display(demographic_table)

# ذخیره
demographic_table.to_csv('table1_demographic.csv', index=False)


# ===================================================
# 2️⃣ جدول آمار توصیفی نمرات
# ===================================================
desc_stats = df[['نمره_بیخوابی', 'نمره_اضطراب', 'نمره_نشخوار']].describe().T
desc_stats = desc_stats[['mean', 'std', 'min', 'max']]
desc_stats.columns = ['میانگین', 'انحراف معیار', 'حداقل', 'حداکثر']
desc_stats.index = ['بی‌خوابی', 'اضطراب', 'نشخوار فکری']

print("\n📊 جدول ۲: آمار توصیفی نمرات")
display(desc_stats)

# ذخیره
desc_stats.to_csv('table2_descriptive.csv')


# ===================================================
# 3️⃣ جدول همبستگی (با سطوح معناداری)
# ===================================================
vars_list = ['نمره_بیخوابی', 'نمره_اضطراب', 'نمره_نشخوار']
corr_matrix = np.zeros((3, 3))
p_matrix = np.zeros((3, 3))

for i, var1 in enumerate(vars_list):
    for j, var2 in enumerate(vars_list):
        corr, p = pearsonr(df[var1], df[var2])
        corr_matrix[i, j] = corr
        p_matrix[i, j] = p

# ساخت دیتافریم
corr_table = pd.DataFrame(corr_matrix, index=vars_list, columns=vars_list)
p_table = pd.DataFrame(p_matrix, index=vars_list, columns=vars_list)

# نمایش همبستگی با ستاره‌های معناداری
corr_with_stars = corr_table.round(3).astype(str)
for i in range(3):
    for j in range(3):
        if p_table.iloc[i, j] < 0.001:
            corr_with_stars.iloc[i, j] += '***'
        elif p_table.iloc[i, j] < 0.01:
            corr_with_stars.iloc[i, j] += '**'
        elif p_table.iloc[i, j] < 0.05:
            corr_with_stars.iloc[i, j] += '*'

print("\n📊 جدول ۳: ماتریس همبستگی (با سطوح معناداری)")
print("*** p<0.001, ** p<0.01, * p<0.05")
display(corr_with_stars)

# ذخیره (همراه با p-valueها)
corr_table.to_csv('table3_correlation.csv')
p_table.to_csv('table3_pvalues.csv')


# ===================================================
# 4️⃣ جدول تحلیل میانجی (ضرایب مسیرها و سوبل)
# ===================================================
# استخراج ضرایب و خطای استاندارد از مدل‌های قبلی
coef_a = model_a.params['نمره_اضطراب']
se_a = model_a.bse['نمره_اضطراب']
p_a = model_a.pvalues['نمره_اضطراب']

coef_b = model_b.params['نمره_نشخوار']
se_b = model_b.bse['نمره_نشخوار']
p_b = model_b.pvalues['نمره_نشخوار']

coef_c = model_c.params['نمره_اضطراب']
se_c = model_c.bse['نمره_اضطراب']
p_c = model_c.pvalues['نمره_اضطراب']

coef_cprime = model_b.params['نمره_اضطراب']
se_cprime = model_b.bse['نمره_اضطراب']
p_cprime = model_b.pvalues['نمره_اضطراب']

# ساخت جدول
mediation_table = pd.DataFrame({
    'مسیر': ['a (اضطراب → نشخوار)', 
             'b (نشخوار → بی‌خوابی با کنترل اضطراب)',
             'c کل (اضطراب → بی‌خوابی)',
             'c\' مستقیم (اضطراب → بی‌خوابی با کنترل نشخوار)'],
    'ضریب': [coef_a, coef_b, coef_c, coef_cprime],
    'خطای استاندارد': [se_a, se_b, se_c, se_cprime],
    'p-value': [p_a, p_b, p_c, p_cprime],
    'معنادار': ['✅' if p < 0.05 else '❌' for p in [p_a, p_b, p_c, p_cprime]]
})

# اضافه کردن نتایج سوبل به عنوان ردیف آخر
sobel_row = pd.DataFrame({
    'مسیر': ['اثر غیرمستقیم (Sobel)'],
    'ضریب': [indirect_effect],
    'خطای استاندارد': [sobel_se],
    'p-value': [p_value],
    'معنادار': ['✅' if p_value < 0.05 else '❌']
})

mediation_table = pd.concat([mediation_table, sobel_row], ignore_index=True)

print("\n📊 جدول ۴: نتایج تحلیل میانجی (روش بارون و کنی + سوبل)")
display(mediation_table)

# ذخیره
mediation_table.to_csv('table4_mediation.csv', index=False)

print("\n✅ همه جداول با موفقیت ذخیره شدند.")
```

```python
# ====================================================
# ۱. ایمپورت کتابخانه‌ها
# ====================================================
import pandas as pd
import numpy as np
import statsmodels.api as sm

# ====================================================
# ۲. خواندن فایل اکسل
# ====================================================
file_path = r'C:\Users\leila.gh\Desktop\Bikhabi.xlsx'
df = pd.read_excel(file_path, sheet_name='Results', header=0)
df = df.dropna(how='all')

# ====================================================
# ۳. شناسایی ستون‌ها (مطابق کدهای قبلی شما)
# ====================================================
insomnia_cols = [
    '1-لطفا شدت مشکل یا مشکلات بی خوابی کنونی ( یعنی دو هفته اخیر) خود را تخمین بزنید.\nالف.مشکل در به خواب رفتن',
    '1-لطفا شدت مشکل یا مشکلات بی خوابی کنونی ( یعنی دو هفته اخیر) خود را تخمین بزنید.\nب.مشکل در تداوم خواب',
    '1-لطفا شدت مشکل یا مشکلات بی خوابی کنونی ( یعنی دو هفته اخیر) خود را تخمین بزنید.\nج. مشکل در زود بیدار شدن',
    '2- میزان رضایت/عدم رضایت شما از الگوی اخیر خوابتان چگونه است؟',
    '3-- فکر می کنید مشکل خواب شما تا چه حد در کارکرد روزانه شما (برای مثال خستگی روزانه، توانایی انجام کار/کارهای روزمره، تمرکز، حافظه، خلق و غیره) تداخل می کند؟',
    '4- فکر می کنید نسبت به دیگران چقدر مشکل خوابیدن شما از نظر تخریب کیفیت زندگی تان قابل توجه است؟',
    '5- \xa0در حال حاضر تا چه اندازه درباره مشکل خواب خود نگران/آشفته هستید؟'
]

anxiety_cols = [
    '1- طی دو هفته گذشته، هر چند وقت یکبار با مشکلات زیر اذیت شده اید؟\n\xa0 ( احساس عصبی بودن، اضطراب یا نزدیک مرز بودن)',
    '2- طی دو هفته گذشته، هر چند وقت یکبار با مشکلات زیر اذیت شده اید؟\n(قادر نبودن به جلوگیری یا کنترل نگرانی)',
    '3-طی دو هفته گذشته، هر چند وقت یکبار با مشکلات زیر اذیت شده اید؟\n\xa0(نگرانی بیش از حد در مورد چیزهای مختلف)',
    '4- طی دو هفته گذشته، هر چند وقت یکبار با مشکلات زیر اذیت شده اید؟\n(سختی در آرام بودن و احساس آرامش)',
    '5-طی دو هفته گذشته، هر چند وقت یکبار با مشکلات زیر اذیت شده اید؟\n( آنقدر بی قرار بودم که برایم آرام نشستن سخت است)',
    '6-طی دو هفته گذشته، هر چند وقت یکبار با مشکلات زیر اذیت شده اید؟\n(به آسانی آزرده خاطر یا تحریک می شوم)',
    '7- طی دو هفته گذشته، هر چند وقت یکبار با مشکلات زیر اذیت شده اید؟\n( احساس ترس به دلیل اینکه ممکن است اتفاق وحشتناکی رخ دهد)'
]

rumination_cols = [
    '1- تفکر فلسفی یا نظری چندان برایم جذاب نیست.',
    '2- واقعا آدم اهل تامل و تفکری نیستم',
    '3- عاشق کاوش در جنبه های درونی خودم هستم',
    '4- مدت ها پس از اینکه بحث و جدلی تمام می شود فکرم دائما به آن باز می گردد',
    '5- وقتی چیزی برایم رخ می دهد، تا مدت ها پس از وقوع آن درباره اش فکر می کنم و به آن متمرکز می شوم',
    '6- وقتم را با فکر کردن درباره چیزهایی که اتفاق افتاده و تمام شده تلف نمی کنم.',
    '7- ذاتا آدم خیلی کنجکاوی در مورد خودم هستم و دوست دارم خودم را بشناسم',
    '8- برایم خیلی راحت است افکار ناخواسته را از ذهنم خارج کنم.',
    '9- عاشق تفکر و تامل درباره ماهیت و معنای امور مختلف هستم.',
    '10- وقت زیادی را به فکر کردن در خصوص لحظه هایی که مایوس و شرمنده شده ام صرف می کنم.',
    '11-تامل کردن درباره خود برایم جالب نیست.'
]

# ====================================================
# ۴. کدگذاری (نگاشت مقادیر متنی به عددی)
# ====================================================
insomnia_map = {
    'هرگز': 0, 'کمی': 1, 'متوسط': 2,
    'شدید': 3, 'بسیار شدید': 4
}
anxiety_map = {
    'اصلا': 0, 'چندین روز': 1,
    'بیش از نیمی از روز': 2, 'تقریبا هر روز': 3
}
rumination_map = {
    'خیلی مخالفم': 0, 'مخالفم': 1,
    'نه موافقم و نه مخالفم': 2,
    'موافقم': 3, 'خیلی موافقم': 4
}

for col in insomnia_cols:
    df[col] = df[col].map(insomnia_map)
for col in anxiety_cols:
    df[col] = df[col].map(anxiety_map)
for col in rumination_cols:
    df[col] = df[col].map(rumination_map)

# ====================================================
# ۵. محاسبه نمرات کل
# ====================================================
df['نمره_بیخوابی'] = df[insomnia_cols].sum(axis=1)
df['نمره_اضطراب'] = df[anxiety_cols].sum(axis=1)
df['نمره_نشخوار'] = df[rumination_cols].sum(axis=1)

# حذف ردیف‌های گمشده در متغیرهای اصلی
df_clean = df[['نمره_بیخوابی', 'نمره_اضطراب', 'نمره_نشخوار']].dropna()

# ====================================================
# ۶. تحلیل میانجی (مسیرهای a, b, c, c')
# ====================================================
X = df_clean['نمره_اضطراب']
M = df_clean['نمره_نشخوار']
Y = df_clean['نمره_بیخوابی']

# مسیر a
model_a = sm.OLS(M, sm.add_constant(X)).fit()
a = model_a.params['نمره_اضطراب']

# مسیر b (با کنترل X)
X_M = pd.DataFrame({'نمره_اضطراب': X, 'نمره_نشخوار': M})
model_b = sm.OLS(Y, sm.add_constant(X_M)).fit()
b = model_b.params['نمره_نشخوار']

# مسیر کل c
model_c = sm.OLS(Y, sm.add_constant(X)).fit()
c = model_c.params['نمره_اضطراب']

# مسیر مستقیم c'
c_prime = model_b.params['نمره_اضطراب']

# ====================================================
# ۷. بوت‌استرپ برای فاصله اطمینان اثر غیرمستقیم
# ====================================================
n_bootstrap = 1000
indirect_effects = []

for _ in range(n_bootstrap):
    sample = df_clean.sample(n=len(df_clean), replace=True)
    X_s = sample['نمره_اضطراب']
    M_s = sample['نمره_نشخوار']
    Y_s = sample['نمره_بیخوابی']
    
    # مدل a در نمونه بوت‌استرپ
    model_a_bs = sm.OLS(M_s, sm.add_constant(X_s)).fit()
    a_bs = model_a_bs.params['نمره_اضطراب']
    
    # مدل b در نمونه بوت‌استرپ
    X_M_bs = pd.DataFrame({'نمره_اضطراب': X_s, 'نمره_نشخوار': M_s})
    model_b_bs = sm.OLS(Y_s, sm.add_constant(X_M_bs)).fit()
    b_bs = model_b_bs.params['نمره_نشخوار']
    
    indirect_effects.append(a_bs * b_bs)

ci_lower = np.percentile(indirect_effects, 2.5)
ci_upper = np.percentile(indirect_effects, 97.5)

# ====================================================
# ۸. چاپ نتایج نهایی
# ====================================================
print("📊 نتایج تحلیل میانجی:")
print(f"مسیر a (اضطراب → نشخوار): ضریب = {a:.4f}")
print(f"مسیر b (نشخوار → بی‌خوابی با کنترل اضطراب): ضریب = {b:.4f}")
print(f"مسیر کل c (اضطراب → بی‌خوابی): ضریب = {c:.4f}")
print(f"مسیر مستقیم c' (اضطراب → بی‌خوابی با کنترل نشخوار): ضریب = {c_prime:.4f}")
print(f"اثر غیرمستقیم (a×b) = {a * b:.4f}")
print(f"فاصله اطمینان ۹۵٪ بوت‌استرپ: [{ci_lower:.4f}, {ci_upper:.4f}]")

if ci_lower < 0 < ci_upper:
    print("❌ نتیجه: اثر غیرمستقیم در سطح ۰٫۰۵ معنادار نیست (صفر در فاصله اطمینان قرار دارد).")
else:
    print("✅ نتیجه: اثر غیرمستقیم در سطح ۰٫۰۵ معنادار است.")
    
```
